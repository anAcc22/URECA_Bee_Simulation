```python
import json
import copy
import scipy
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
import matplotlib.ticker as ticker

sns.set_theme(font='Fira Code', font_scale=0.8, style='ticks')
# sns.set_style({'grid.linestyle': ':'})
```

```python
def get_filename(crawl, climb, size, mass, count):
    return f'../data/crawl_{crawl}_climb_{climb}_size_{size}_mass_{mass}_cnt_{count}.json';
```

```python
def power(x, a, C):
    return C * pow(x, a)
```

## **Part I: Varying Bee Count (crawl_rate = climb_rate = 0.4)**

```python
palette = sns.color_palette('crest', n_colors=8)
palette.reverse()
```

```python
crawl, climb, size, mass = '0.4', '0.4', '0', '0'
counts = [100, 150, 200, 250, 300, 350, 400, 450]
```

```python
rows = []

for count in counts:
    filename = get_filename(crawl, climb, size, mass, count)
    with open(filename, 'r') as file:
        data = json.load(file)
    for graph in data['widthGraphs']:
        for point in graph:
            rows.append([point['x'], point['y'], count])
            
df = pd.DataFrame(rows, columns=['x', 'y', 'Count'])

plt.figure(figsize=(7, 6))
ax = sns.lineplot(x='x', y='y', data=df, hue='Count', palette=palette, marker='o', markersize=5, alpha=0.7)
ax.set(xlabel='z (millipixels)', ylabel='Diameter (millipixels)')
sns.despine(ax=ax, offset=10, trim=True)

plt.show()
```

## **Part IV: Varying both climb_rate and crawl_rate (bee_count = 100)**

```python
def get_boundary(a, n = 10):
    max_x = 0

    for p in a:
        max_x = max(max_x, p['x'])
        
    chunk =  max_x / n;
    best = [0 for i in range(n)]

    # filler = [{'x': 0, 'y': 0}]
    
    for p in a:
        h = min(n-1, math.floor(p['x'] / chunk))
        best[h] = max(best[h], p['y'])

    def non_zero(p):
        return p['y'] != 0
        
    ans = [{ 'x': float(chunk*i) + chunk / 2, 'y': best[i] } for i in range(n)]

    fx = -chunk / 2
    lx = 1 + chunk / 2

    return [{'x': fx, 'y': 0}] + ans + [{'x': lx, 'y': 0}]
```

```python
def auto_scale(a, n = 10):
    new_a = copy.deepcopy(a)
    
    min_x = min([p['x'] for p in a])
    min_y = min([p['y'] for p in a])

    for p in new_a:
        p['x'] -= min_x
        p['y'] -= min_y

    scale = max([p['x'] for p in new_a])

    for p in new_a:
        p['x'] /= scale
        p['y'] /= scale

    return new_a
```

```python
def get_curvature(points, n=10):
    points = get_boundary(auto_scale(points), n)
    points = [np.array([p['x'], p['y'], 0]) for p in points]
    
    if len(points) < 3:
        return []

    tangs = np.diff(points, axis=0)
    curvs = []

    for i in range(len(tangs)-1):
        v1 = tangs[i]
        v2 = tangs[i+1]
        
        dotp = np.dot(v1, v2)
        prod_mag = np.linalg.norm(v1) * np.linalg.norm(v2)
        
        if prod_mag == 0:
            cos_theta = 1.0
        else:
            cos_theta = dotp / prod_mag
            
        curvature = 2 * (1 - cos_theta)

        cross_prod = v1[0] * v2[1] - v1[1] * v2[0]

        if cross_prod > 0:
           curvature *= -1
            
        curvs.append(curvature)
        
    return np.mean(curvs)
```

```python
def display_curvature_array(points, ax, n=10):
    points = get_boundary(auto_scale(points), n)
    points = [np.array([p['x'], p['y'], 0]) for p in points]
    
    if len(points) < 3:
        return []

    tangs = np.diff(points, axis=0)
    curvs = []

    for i in range(len(tangs)-1):
        v1 = tangs[i]
        v2 = tangs[i+1]
        
        dotp = np.dot(v1, v2)
        prod_mag = np.linalg.norm(v1) * np.linalg.norm(v2)
        
        if prod_mag == 0:
            cos_theta = 1.0
        else:
            cos_theta = dotp / prod_mag
            
        curvature = 2 * (1 - cos_theta)

        cross_prod = v1[0] * v2[1] - v1[1] * v2[0]

        if cross_prod > 0:
           curvature *= -1
            
        curvs.append(curvature)
        
    rows = []
    
    for x in range(len(curvs)):
        rows.append([x/(len(curvs)-1), curvs[x]])
        
    df = pd.DataFrame(rows, columns=['x', 'y'])
    main_color = '#222222'
    saturation_scale = 4
    
    df['shade'] = df['y'].apply(lambda y: 1 if y > 0 else 0)
    df['saturation'] = df['y'].apply(lambda y: abs(y) / saturation_scale + 0.2)

    ax.plot(df['x'], df['y'], color=main_color, linewidth=1, zorder=1)
    norm_original = Normalize(vmin=0.0, vmax=1.0)

    df_p = df[df['shade'] == 1]
    scatter_p = ax.scatter(
        df_p['x'],
        df_p['y'],
        c=df_p['saturation'],
        cmap='Greens',
        s=200 * df_p['saturation'],
        zorder=2,
        linewidth=1,
        edgecolor=main_color,
        norm=norm_original,
    )

    df_n = df[df['shade'] == 0]
    scatter_n = ax.scatter(
        df_n['x'],
        df_n['y'],
        c=df_n['saturation'],
        cmap='Reds',
        s=160 * df_n['saturation'],
        zorder=2,
        linewidth=1,
        edgecolor=main_color,
        norm=norm_original,
    )
    
    ax.set_xlim([-0.1, 1.1])
    ax.set_ylim([-2.5, 2.5])
```

```python
def display_grid():
    scale = int(count)/100*0.2 + 1
    rows = []
    for crawl in crawls:
        for climb in climbs:
            filename = get_filename(crawl, climb, size, mass, count)
            with open(filename, 'r') as file:
                data = json.load(file)
            max_x = max([point['x'] for point in data['points']])
            min_x = min([point['x'] for point in data['points']])
            halved = (max_x - min_x) / 2
            for point in data['points']:
                rows.append([point['x'] - halved, -point['y'], crawl, climb])
    
    df = pd.DataFrame(rows, columns=['x', 'y', 'Crawl Rate', 'Climb Rate'])
    
    g = sns.FacetGrid(df, col='Crawl Rate', row='Climb Rate', margin_titles=True, despine=False, height=1.5, xlim=[-500*scale, 500*scale], ylim=[-1000*scale, 0])
    g.map(sns.scatterplot, 'x', 'y', marker='o', s=5, alpha=0.7, color='#b08020')
    g.set_titles(col_template='{col_name}', row_template='{row_name}', size=14)
    
    g.set(xticks=[])
    g.set(yticks=[])
    g.set(xlabel=None)
    g.set(ylabel=None)
    g.figure.supxlabel('Crawl Rate', fontsize=14, y=0.96)
    g.figure.text(
        x=0.98,
        y=0.5,
        s='Climb Rate',
        fontsize=14,
        rotation=270,
        verticalalignment='center',
        horizontalalignment='center'
    )
    
    g.figure.subplots_adjust(wspace=0, hspace=0, top=0.92, right=0.92)
    
    plt.show()
```

```python
def display_curvatures():
    fig, axes = plt.subplots(
        len(climbs),
        len(crawls),
        figsize=(10, 10),
        sharex=True,
        sharey=True,
    )
    
    for i, climb in enumerate(climbs):
        for j, crawl in enumerate(crawls):
            ax = axes[i, j]
            filename = get_filename(crawl, climb, size, mass, count)
            with open(filename, 'r') as file:
                data = json.load(file)
            display_curvature_array(data['points'], ax)
    
            if i == 0:
                ax.set_title(f'{crawl}', size=14)
            if j == 0:
                ax.tick_params(left=True, right=False, top=False, bottom=False)
            else:
                ax.tick_params(left=False, right=False, top=False, bottom=False)
    
            if j == len(crawls) - 1:
                ax.yaxis.set_label_position('right')
                ax.set_ylabel(f'{climb}', rotation=270, labelpad=14, size=14)
            else:
                ax.set_ylabel(None)
            
            ax.set_xticks([])
            ax.set_xlabel(None)
            
    fig.supylabel('Curvature', fontsize=14, x=0.04)
    fig.supxlabel('Crawl Rate', fontsize=14, y=0.92)
    fig.text(
        x=0.95,
        y=0.5,
        s='Climb Rate',
        fontsize=14,
        rotation=270,
        verticalalignment='center',
        horizontalalignment='center'
    )
    
    fig.subplots_adjust(wspace=0, hspace=0, left=0.1, right=0.9)
```
